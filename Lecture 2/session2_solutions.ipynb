### 2 · `.loc[rows, columns]` — by **label**
The index here is 0, 1, 2 … so labels *look* like positions. Slices **include** the stop label.

```python
df.loc[0, "fare"]                          # one cell
```

```python
df.loc[0:4, "fare":"total"]                # rows 0..4 AND columns fare..total -> stop included
```

```python
df.loc[[0, 10, 20], ["pickup_zone", "total"]]   # lists of labels
```

### 3 · `.iloc[rows, columns]` — by **position**
Slices **exclude** the stop. Negative positions count from the end.

```python
df.iloc[0, 4]                    # first row, 5th column (fare)
```

```python
df.iloc[0:5, :]                  # positions 0..4
```

```python
df.iloc[[0, 2, 4], [3, 5]]       # rows 0, 2, 4 · columns distance, tip
```

```python
df.iloc[-3:]                     # the last 3 rows
```

### 4 · After sorting, labels stay, positions move

```python
s = df.sort_values("total", ascending=False)
print("iloc[0] ->", s.iloc[0]["total"], "(the most expensive ride)")
print("loc[0]  ->", s.loc[0]["total"], "(the ride with label 0, wherever it is now)")
s.head(3)
```

### 5 · Slicing: `start : stop : step`

```python
df.iloc[::1000, :4]          # every 1000th ride, first 4 columns
```

```python
df.iloc[:, -4:].tail(2)      # last 4 columns, last 2 rows
```

```python
df["fare"].iloc[::-1].head(3)   # step -1 reverses
```

### 6 · Boolean indexing
condition → **mask** (True/False per row) → `df[mask]` keeps the True rows

```python
mask = df["total"] > 100
mask.head()
```

```python
print("rows matching:", mask.sum())
print("share:", round(mask.mean(), 4))
df[mask][["pickup_zone", "dropoff_zone", "distance", "total"]]
```

### 7 · Combine: `&` and · `|` or · `~` not — each test in `( )`

```python
green_long = (df["color"] == "green") & (df["distance"] > 10)
df[green_long].shape
```

```python
no_manhattan = ~(df["pickup_borough"] == "Manhattan")
df.loc[no_manhattan, "pickup_borough"].value_counts(dropna=False)
```

### 8 · Every kind of condition

```python
df[df["pickup_borough"].isin(["Bronx", "Brooklyn"])].shape     # in a list
```

```python
df[df["tip"].between(5, 10)].shape                           # in a range (both ends included)
```

```python
df[df["payment"].isna()].head(3)                            # missing values
```

```python
df.nlargest(3, "tip")[["pickup_zone", "fare", "tip"]]          # top n
```

### 9 · `.query()` — the same filter, as a sentence

```python
a = df[(df["color"] == "green") & (df["distance"] > 10)]
b = df.query("color == 'green' and distance > 10")
a.equals(b)
```

```python
limit = 50
df.query("total > @limit and payment == 'cash'")[["pickup_zone", "total"]].head()
```

### 10 · Copy-on-Write (pandas ≥ 3.0)
Chained assignment never changes `df` → you get a `ChainedAssignmentError` warning. Use **one** `.loc`.

```python
tmp = df.copy()
tmp[tmp["payment"] == "cash"]["tip"] = -1          # WRONG: two [ ] -> warning, nothing changes
print((tmp["tip"] == -1).sum())
```

```python
tmp.loc[tmp["payment"] == "cash", "tip"] = -1     # RIGHT: one .loc
print((tmp["tip"] == -1).sum())
```

```python
sub = df[df["color"] == "green"]
sub["fare"] = 0                                      # changes sub only: df is untouched
print(sub["fare"].max(), df.loc[df["color"] == "green", "fare"].max())
```

### Block E — Small case studies (★★★, ~13 min)
Each question is a short story. Build the masks one at a time, then combine them.

**★★★ · Q19 — Strange records.**
Before publishing statistics, the team looks for records that make no sense.
1. How many rides have **0 passengers**?
2. How many rides have a **distance of 0**?
3. How many rides have **at least one** of the two problems?
   Is it the sum of 1 and 2? (*Hint:* count the rides with **both** problems.)

```python
no_passengers = df["passengers"] == 0
no_distance = df["distance"] == 0
print(no_passengers.sum(), no_distance.sum())
print((no_passengers | no_distance).sum())
print((no_passengers & no_distance).sum())
# 96 rides with 0 passengers, 51 with distance 0, but only 146 (not 147) with at least one problem:
# 1 ride has both problems, and | counts it once.
```

**★★★ · Q20 — The JFK flat fare.**
Yellow cabs charge a **flat fare of \$52** from Manhattan to JFK Airport (tolls, tip and extras come on top).
1. How many rides start in Manhattan (`pickup_borough`) **and** end at `"JFK Airport"` (`dropoff_zone`)?
2. Look at the `fare` of these rides: how many times does each value appear? Is it always \$52?

```python
to_jfk = (df["pickup_borough"] == "Manhattan") & (df["dropoff_zone"] == "JFK Airport")
print(to_jfk.sum())
df.loc[to_jfk, "fare"].value_counts()
# 40 rides from Manhattan to JFK, and every fare is exactly $52: the rule holds.
```

**★★★ · Q21 — Do cash customers tip?**
1. Count the rides paid in **cash** **with a tip above 0**.
2. Compute the average tip of **credit card** rides and of **cash** rides.

```python
cash = df["payment"] == "cash"
card = df["payment"] == "credit card"
print((cash & (df["tip"] > 0)).sum())
print(round(df.loc[card, "tip"].mean(), 2), round(df.loc[cash, "tip"].mean(), 2))
# 0 cash rides have a tip: card tips average $2.78, cash tips $0.
# Cash tips go straight into the driver's hand, so they are never recorded in the data.
```
